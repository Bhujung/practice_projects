```python
# inheritance
class grandfather:
    def skill1(self):
        print('I am farmer.')

class father(grandfather):
    def skill2(self):
        print('I am doctor.')

class son(father):
    def skill3(self):
        print('I am a singer')

s1 = son()
s1.skill1()
s1.skill2()
s1.skill3()
```

```python
class father:
    def skill(self):
        print('I am Farmer')
class son(father):
    def skill(self):
        super().skill() # super le chai parent ko skill lai dekhaideu baneko ho 
        print('I am Gamer')
s1 = son()
s1.skill()
```

```python
# constructor super class
class employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class developer(employee):
    def __init__(self, name, salary, language):
        super().__init__(name, salary)
        self.language = language
        print('Developer details stored.')

    def show_details(self):
        print(f'Name : {self.name} \nSalary : {self.salary} \nLanguage : {self.language}')

d1 = developer('Ram', 312321, 'Nepali')
d1.show_details()
```
