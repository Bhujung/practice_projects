```python
# Personal Dashboard 
name = input('Enter a name: ')

while True:
    try:
        age = int(input("Enter your age: "))
        if age < 0:
            print("Age can't be negative! Try again.")
            continue
        break
    except ValueError:
        print("That's not a number! Try again.")

city = input('Enter a city: ')
job_title = input('Enter your job title: ')

while True:
    try:
        monthly_income = int(input('Enter your Monthly Salary: '))
        if monthly_income < 0:
            print('Monthly income cannot be negative! Try again.')
            continue
        break
    except ValueError:
        print('That is not the number. Try again')


yearly_income = monthly_income * 12

print("\n") 
print("=" * 40)
print("   PERSONAL DATA DASHBOARD")
print("=" * 40)
print("Name :", name)
print("Age :", age)
print("City :", city)
print("Job Title :", job_title)
print("Monthly Income : $" + str(monthly_income))
print("Yearly Income : $" + str(yearly_income))
print("=" * 40)
```

```python
# Text data cleaner
print("=" * 45)
print("   WELCOME TO TEXT DATA CLEANER")
print("=" * 45)

raw_name = input('Enter your raw name: ')
raw_email = input('Enter your raw email:')
raw_phone = input('Enter your raw phone number')
raw_city = input('Enter a raw city: ')

clean_name = raw_name.strip().title()
clean_email = raw_email.strip().lower()
clean_city = raw_city.strip().title()
clean_phone = raw_phone.replace('-','').replace(' ','')

print("Clean Data Report")
print('='*45)
print('Original Name: ', repr(raw_name))
print('Cleaned Name: ',(clean_name))
print('Original Email: ',repr(raw_email))
print('Cleaned Email: ',(clean_email))
print('Original Phone: ',repr(raw_phone))
print('Cleaned Phone: ',(clean_phone))
print('Original City: ',repr(raw_city))
print('Cleaned City: ',(clean_city))
print("=" * 45)

name_words = clean_name.split()
words = len(name_words)
print(f'The length of the words is {words}')
```

```python
# Sales List Manager
print('='*45)
print('Welcome to the sales list manager')
print('='*45)

sales = [] # created sales list to store sales data

num_days = int(input('Enter number of days'))

for day in range(1, num_days + 1):
    while True:
        try:
            amount = float(input(f'Enter sales of the day {day}$'))
            if amount > 0:
                sales.append(amount)
                break
            else:
                print('sales cannot be negative')
        except ValueError:
            print('Please enter a number')
total_sales = sum(sales)
average_sales = total_sales/len(sales)
max_sales = max(sales) # max shows the maximum number of sales
min_sales = min(sales) # min shows the minimum number of sales

print('Sales Report')
print('Sales',sales)
print('Total sales : ${:,.2f}'.format(total_sales))
print('Average sales : ${:,.2f}'.format(average_sales))
print('Total sales : ${:,.2f}'.format(total_sales))
print('Best sales : ${:,.2f}'.format(max_sales))
print('Worst Sales : ${:,.2f}'.format(min_sales))

best_day_number= sales.index(max_sales)+1
print(f'The best day of sales is {best_day_number}')
```

```python
# student gradebook

print('='*45)
print('         Welcome to student gradebook')
print('='*45)

print('-> All students and their grades')
print('-'*45)
gradebook = {
    'Budhha' : 55,
    'Kanchan' : 65,
    'Prabin' : 45,
    'Anil' : 50,
    'Suresh' : 45
}

for name,grade in gradebook.items():
    status = 'pass' if grade >= 40 else 'Fail'
    print(f'{name}: {grade}({status})')

search = input('\nEnter a student name: ').strip().title()

if search in gradebook:
    print(f'{search} grade is {gradebook[search]}')
else:
    print('The student is not registered yet')

print('-'*45)

new_student = input('\nEnter a new student name: ').strip().title()
while True:
    try :
        new_grade = int(input(f'\nEnter a new grade of {new_student} :'))
        if 100 > new_grade >= 0:
            break
        else :
            print('Grade must me between 0 and 100')
    except ValueError:
        print('Please enter a number')
gradebook[new_grade] = new_grade
print(f'\nGrade of {new_student} is {new_grade}')

print('\nUpdated Gradebook')
for name,grade in gradebook.items():
    print(f'{name} : {grade}')
```
